# Zomato Restaurant Dataset Analysis

## Overview
This notebook analyzes a Zomato restaurant dataset containing 148 records. The dataset includes restaurant-level information such as whether online ordering is available, table booking availability, ratings, votes, approximate cost for two people, and restaurant category.

## Dataset Description
The dataset contains the following columns:

- name: Restaurant name
- online_order: Whether the restaurant accepts online orders (Yes/No)
- book_table: Whether table booking is available (Yes/No)
- rate: Restaurant rating
- votes: Number of user votes received
- approx_cost(for two people): Estimated cost for two people
- listed_in(type): Restaurant category/type such as Buffet, Cafes, Dining, and other

## Objective
The main purpose of this analysis is to understand:

- Which restaurant types are most common
- Which categories receive the highest number of votes
- How ratings differ between restaurants that do and do not offer online ordering
- Cost distribution among restaurants
- Relationship between restaurant type and online order availability

## Data Cleaning and Preprocessing
Before analysis, the rating column was cleaned and converted into numeric values for easier comparison and visualization.

The preprocessing step included:
- Converting the rating values to strings
- Splitting the values at the "/" symbol
- Keeping the value before the slash
- Converting the result to float

This ensured that rating values such as "4.1/5" were treated as numeric values like 4.1.

## Key Insights
From the available data:
- The most common restaurant categories include Buffet, Cafes, Dining, and other
- Dining restaurants received the highest total votes
- Restaurants that accept online orders show a different rating distribution compared to those that do not
- Approximate cost for two people varies widely across categories
- Some categories have more restaurants with online ordering enabled than others

## Analysis Performed
The notebook includes the following analyses:

### 1. Restaurant Type Distribution
A count plot was created to display how many restaurants belong to each listed_in(type) category.

### 2. Votes by Restaurant Type
The total votes for each restaurant type were grouped and visualized using a line chart.

### 3. Rating Distribution
A histogram of the rate column was plotted to understand the spread of restaurant ratings.

### 4. Cost Distribution
The approx_cost(for two people) column was analyzed to observe how frequently different price ranges appear.

### 5. Online Orders vs Ratings
A box plot was used to compare restaurant ratings for:
- Restaurants with online orders
- Restaurants without online orders

### 6. Heatmap Analysis
A pivot table was created to show the relationship between:
- Restaurant type
- Online order availability

This was then visualized using a heatmap to identify patterns in the data.

## Summary
This dataset provides useful insights into restaurant trends, customer engagement, pricing, and service offerings. It helps us understand which restaurant categories are more popular, how ratings vary, and how online ordering affects restaurant performance.

## Final Conclusion
The Zomato dataset is a practical example of exploratory data analysis (EDA). It demonstrates how data cleaning, grouping, visualization, and summarization can be used to derive valuable business insights from restaurant data.

### ⭐ Handling Restaurant Ratings

**Objective:** To clean and convert restaurant ratings into numeric values for analysis.

- 🔤 **`str(value)`**: Converts the rating into a string.
- ✂️ **`split('/')`**: Splits the rating wherever `/` appears.
- 🎯 **`value[0]`**: Selects the part before `/`.
- 🔢 **`float(value)`**: Converts the selected value into a decimal number.
- 🔄 **`apply(handleRate)`**: Applies the function to every value in the `rate` column.

```python
def handleRate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

df['rate'] = df['rate'].apply(handleRate)
```

### 📈 Restaurant Type vs Total Votes

**Objective:** To analyze and visualize the total votes received by different types of restaurants in the Zomato dataset.

- 🗂️ **`groupby()`**: Groups the data according to restaurant type.
- 🗳️ **`sum()`**: Calculates the total votes for each restaurant type.
- 📊 **`pd.DataFrame()`**: Stores the grouped vote totals in a DataFrame.
- 📉 **`plt.plot()`**: Creates a line graph to visualize the total votes for each restaurant type.
- 🟢 **`c='green', marker='o'`**: Displays a green line with circular markers.
- 🏷️ **`plt.xlabel()`**: Labels the x-axis as *Type of Restaurant*.
- 🗳️ **`plt.ylabel()`**: Labels the y-axis as *Votes*.
- 🔄 **`plt.xticks(rotation=45)`**: Rotates the x-axis labels by 45 degrees for better readability.
- 📐 **`plt.tight_layout()`**: Adjusts the graph layout to prevent labels from overlapping or getting cut off.
- 👁️ **`plt.show()`**: Displays the graph.

```python

import pandas as pd
import matplotlib.pyplot as plt

grouped_data = df.groupby('listed_in(type)')['votes'].sum()

result = pd.DataFrame({'votes': grouped_data})

plt.plot(result.index, result['votes'], c='green', marker='o')

plt.xlabel("Type of Restaurant", c="red", size=12)
plt.ylabel("Votes", c="red", size=12)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 🔥 Heatmap of Restaurant Types and Online Orders

**Objective:** To analyze the relationship between restaurant types and online order availability using a heatmap.

- 📊 **`df.pivot_table()`**: Creates a summary table showing the number of restaurants for each restaurant type and online order category.
- 🗂️ **`index='listed_in(type)'`**: Groups the data by restaurant type.
- 🛒 **`columns='online_order'`**: Separates the data according to online order availability.
- 🔢 **`aggfunc='size'`**: Counts the number of records in each group.
- 🔄 **`fill_value=0`**: Replaces missing combinations with zero.
- 🎨 **`sns.heatmap()`**: Visualizes the summary table using different colors.
- 🔍 **`annot=True`**: Displays the numerical values inside the heatmap.
- 🌈 **`cmap="YlOrRd"`**: Uses a yellow-orange-red color scheme.
- 🔢 **`fmt='d'`**: Displays the annotated values as integers.
- 🏷️ **`plt.title()`**: Sets the graph title to *Heatmap*.
- 📌 **`plt.xlabel()`**: Labels the x-axis as *Online Order*.
- 📌 **`plt.ylabel()`**: Labels the y-axis as *Listed In (Type)*.
- 👁️ **`plt.show()`**: Displays the heatmap.

```python
pivot_table = df.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
sns.heatmap(pivot_table, annot=True, cmap="YlOrRd", fmt='d')
plt.title("Heatmap")
plt.xlabel("Online Order")
plt.ylabel("listed In (Type)")
plt.show()
```
